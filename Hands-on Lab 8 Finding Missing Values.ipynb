The functions below will download the dataset into your browser:



```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### 2. Explore the Dataset
##### Task 1: Display basic information and summary statistics of the dataset.


```python
df.info()
df.describe()
```

### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
# Count missing values in each column
missing_values = df.isnull().sum()

print(missing_values)
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 8))

sns.heatmap(df.isnull(), cbar=False, yticklabels=False)

plt.title('Missing Values Heatmap')
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
# Count missing values in the Employment column
missing_employment = df['Employment'].isnull().sum()

print("Number of missing rows in Employment:", missing_employment)
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
# Find the most frequent value in Employment
most_frequent_employment = df['Employment'].mode()[0]

print("Most frequent Employment value:", most_frequent_employment)
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
# Find the most frequent value
most_frequent = df['Employment'].mode()[0]

# Fill missing values with the most frequent value
df['Employment'] = df['Employment'].fillna(most_frequent)

# Verify
print("Most frequent Employment value:", most_frequent)
print("Missing values after imputation:", df['Employment'].isnull().sum())
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
# Display employment counts
print(df['Employment'].value_counts())
```

```python
import matplotlib.pyplot as plt

# Count each Employment category
employment_counts = df['Employment'].value_counts()

labels = [label[:30] + '...' if len(label) > 30 else label
          for label in employment_counts.index]

plt.figure(figsize=(10, 6))
plt.bar(labels, employment_counts.values)

plt.title('Distribution of Employment After Imputation')
plt.xlabel('Employment')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
